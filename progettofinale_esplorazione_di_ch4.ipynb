# **CREAZIONE DI FUNZIONI UTILI**

```python


# FUNZIONE CHE RESTITUISCE SOLO PAESI DELL'UNIONE EUROPEA DA APPLICARE A TUTTI I DF

def filter_eu_countries(df):
    """
    Restituisce un nuovo Dataframe con solo i paesi dell'Unione Europea

    Args:
        df (DataFrame): Il DataFrame da filtrare

    Returns:
        DataFrame: Nuovo DataFrame con solo i paesi dell'Unione Europea
    """

    return df[df['Country'].isin(EU_COUNTRIES)].reset_index(drop=True)


#FUNZIONE CHE CONVERTE TUTTI I VALORI DELLE COLONNE NUMERICHE IN UNA LISTA DI VALORI DI NUMERICI

def convert_values_to_numeric(df):
    """
    Converte i valori di tutte le colonne di un DataFrame in tipo numerico.
    I valori non convertibili verranno sostituiti con NaN.

    Args:
        df(DataFrame): Il DataFrame da analizzare.

    Returns:
        DataFrame: Una copia del DataFrame con valori numerici convertiti.
        list: Lista delle colonne convertite.
    """
    # Creo una copia per evitare modifiche al DataFrame originale
    temp_df = df.copy()
    numeric_cols = []

    # Faccio una iterazione su tutte le colonne e le converte in numeriche
    for col in temp_df.columns:
        if col != 'Country':
            temp_df[col] = pd.to_numeric(temp_df[col], errors='raise')
            numeric_cols.append(col)

    return temp_df, numeric_cols

# funzione chiamata calculate_emissions che calcola le emissioni totali e medie

def calculate_emissions(df):
    """
    Calcola le emissioni totali e medie di uno specifico Gas Serra.

    Args:
        df(DataFrame): Il DataFrame da analizzare.

    Returns:
        DataFrame: Nuovo DataFrame con le emissioni totali e medie calcolate.
    """

    # Converte i valori delle colonne in numerici e ottengo i nomi delle colonne convertite
    # Passa il DataFrame 'df' ricevuto dalla funzione a 'convert_values_to_numeric'
    processed_df, numeric_cols = convert_values_to_numeric(df) # qui viene utilizzata la funzione precedente! <<<

    # Calcolo le emissioni totali e medie sul DataFrame elaborato
    processed_df["Total_Emissions"] = processed_df[numeric_cols].sum(axis=1)
    processed_df["Average_Emission"] = processed_df[numeric_cols].mean(axis=1)

    return processed_df

# L’obiettivo di questa funzione è quello di preparare il grafico a linea

"""

 Nello specifico:
Effettuiamo il setup della figura, indicando una dimensione di 18' in larghezza e 8' in altezza, in modo che il grafico sia largo e facile da
leggere;
Iteriamo poi per ogni Paese nel DataFrame
  # per ogni Paese, filtriamo il DataFrame per selezionare solo le righe corrispondenti a quel Paese;
Effettuiamo il setup della linea:
df.columns[1:-2] seleziona tutte le colonne che rappresentano gli anni;
country_data.iloc[0, 1:-2] estrae i valori delle emissioni per il Paese ciclato lungo tutti gli anni;
plt.plot crea una linea sul grafico, utilizzando gli anni sull’asse delle X e le emissioni sull’asse delle Y.
Infine, customizziamo il grafico con i vari dettagli

"""

def plot_trend_by_country(df, gas_name, save_path=None):
    plt.figure(figsize=(18, 8))
    for country in df['Country']:
        country_data = df[df['Country'] == country]
        plt.plot(df.columns[1:-2], country_data.iloc[0, 1:-2], label=country)

    plt.title(f"{gas_name} Emission Trends by Country ", fontsize = 18)
    plt.xlabel("Years", fontsize = 14)
    plt.ylabel("Emissions (tons)", fontsize = 14)
    plt.legend(title = "Country", bbox_to_anchor = (1.05, 1), loc = 'upper left')
    plt.tight_layout()

    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')

    plt.show()

# FUNZIONE CHE RESTITUISCE SOLO PAESI DELL'UNIONE EUROPEA DA APPLICARE A TUTTI I DF

# LA TENGO COSI PERCHE' GIA' UTILIZZATA IN PRECEDENZA MA UTILE DA RICOPIARE IN FUTURO

"""
def filter_eu_countries(df):

    Restituisce un nuovo Dataframe con solo i paesi dell'Unione Europea

    Args:
        df (DataFrame): Il DataFrame da filtrare

    Returns:
      DataFrame: Nuovo DataFrame con solo i paesi dell'Unione Europea


    return df[df['Country'].isin(EU_COUNTRIES)].reset_index(drop=True)

"""

# Questa funzione riesce a filtrare i dati dal 2005 in poi

def get_2005_data_and_above(df):

  """
Restituisce un nuovo Dataframe con solo i dati dal 2005 in poi

Args:
  df (DataFrame): Il DataFrame da filtrare

Returns:
  DataFrame: Nuovo DataFrame con solo i dati dal 2005

  """

  return list(df.loc[:, '2005':])

```

```python
def format_num(x):

    if pd.isna(x):

        return "(N/A)"

    if x >= 1_000_000:

        return f"{x/1_000_000:.1f}M"

    if x >= 1_000:

        return f"{x/1_000:.0f}K"
    return f"{x:.0f}"

# Prepare dataframes with consistent column names for merging
renewable_summary = df_renewable_mean_for_merge[['Country', 'Renewable_Electricity_Mean']].copy()
renewable_summary.rename(columns={'Renewable_Electricity_Mean': 'media_rinnovabili'}, inplace=True)

co2_summary = filtered_co2_2005_FULL[['Country', 'Total_Emissions']].copy()
co2_summary.rename(columns={'Total_Emissions': 'co2_totale'}, inplace=True)

ghg_summary = filtered_ghg_2005_FULL[['Country', 'Total_Emissions']].copy()
ghg_summary.rename(columns={'Total_Emissions': 'ghg_totale'}, inplace=True)

# Merge the summary dataframes to create the 'summary' table
summary = pd.merge(renewable_summary, co2_summary, on='Country', how='inner')
summary = pd.merge(summary, ghg_summary, on='Country', how='inner')

# Rename 'Country' to 'country' for consistency with how it's used later in the cell
summary.rename(columns={'Country': 'country'}, inplace=True)

# Top 5 per media rinnovabili
top5 = summary.sort_values("media_rinnovabili", ascending=False).head(5).copy()

# Bottom 5 per media rinnovabili
bottom5 = summary.sort_values("media_rinnovabili", ascending=True).head(5).copy()

# Formattazione

for t in [top5, bottom5]:
    t["media_rinnovabili"] = (t["media_rinnovabili"]).round(1).astype(str) + "%"
    t["co2_totale"] = t["co2_totale"].apply(format_num)
    t["ghg_totale"] = t["ghg_totale"].apply(format_num)

    # Define a helper function to create the 'riga' string for each row

    def create_riga_for_row(row):
        ghg_emission_display = row["ghg_totale"]
        # If the formatted GHG total is '0' or '(N/A)', change the display to '_N/A_'
        if ghg_emission_display == '0' or ghg_emission_display == '(N/A)':
            ghg_emission_display = "N//A"

        return (
            "  |  Media Rinnovabili " + row["media_rinnovabili"]
            + "  |  CO2 Totale " + row["co2_totale"]
            + "  |  GHG Totale " + ghg_emission_display
        )
    # Apply the helper function to each row to create the 'riga' column
    t["riga"] = t.apply(create_riga_for_row, axis=1)

print("TOP 5 per media rinnovabili",'\n')
print(top5[["country", "riga"]].to_string(index=False))
print('\n')
print("\nBOTTOM 5 per media rinnovabili",'\n')
print(bottom5[["country", "riga"]].to_string(index=False))

print('\n')
```
